## Creating DataFrames in Pandas 
     

### 1.Import Pandas

```python
import pandas as pd
```

### 2.Create a DataFrame from a Dictionary

```python
student_data = {
    "Name": ["Raja", "Arun", "Priya", "Kavin"],
    "Age": [21, 22, 20, 23],
    "Department": ["CSE", "ECE", "IT", "EEE"]
}

df = pd.DataFrame(student_data)

print(df)
```

### 3: Create a DataFrame from a List

```python
numbers = [[1, 10], [2, 20], [3, 30], [4, 40]]

df = pd.DataFrame(numbers, columns=["ID", "Marks"])

print(df)
```

### 4: Create a DataFrame from a List of Dictionaries

```python
employees = [
    {"Name": "Raja", "Salary": 35000},
    {"Name": "Arun", "Salary": 42000},
    {"Name": "Priya", "Salary": 39000}
]

df = pd.DataFrame(employees)

print(df)
```

### 5: Create a DataFrame with a Custom Index

```python
student_data = {
    "Name": ["Raja", "Arun", "Priya"],
    "Marks": [85, 92, 88]
}

df = pd.DataFrame(student_data, index=["S1", "S2", "S3"])

print(df)
```

### 6: Display DataFrame Information

```python
print(df.shape)
print(df.columns)
print(df.index)
```
