# 📘 22 - Apply, Map, and Lambda FunctionsSometimes built-in Pandas functions aren't enough and I need to apply my own logic to a column. I practice `apply()`, `map()`, and lambda functions here.

## Step 1: Import Pandas

```python
import pandas as pd
```

## Step 2: Create a Sample Dataset

```python
df = pd.DataFrame({'Name': ['Asha','Ravi','Kiran'], 'Score': [85, 45, 92]})df
```

## Step 3: Apply a Lambda Function to a Column

```python
df['Grade'] = df['Score'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')df
```

`apply()` runs the function on every value in the column.

## Step 4: Map Values Using a Dictionary

```python
grade_points = {'Pass': 1, 'Fail': 0}df['GradePoint'] = df['Grade'].map(grade_points)df
```

`map()` is useful for simple value-to-value substitutions.

## Step 5: Apply a Function Across Rows

```python
df['Summary'] = df.apply(lambda row: f"{row['Name']} scored {row['Score']}", axis=1)df
```

`axis=1` tells `apply()` to work row by row instead of column by column.

## Step 6: Define and Reuse a Named Function

```python
def score_band(score):    if score >= 90:        return 'Excellent'    elif score >= 50:        return 'Average'    return 'Needs Improvement'df['Band'] = df['Score'].apply(score_band)df
```

## SummaryI practiced transforming column values with lambda functions using `apply()`, substituting values with `map()`, and applying logic row-wise with `axis=1`.

## Conclusion`apply()`, `map()`, and lambda functions give me the flexibility to add custom logic that built-in Pandas methods don't directly cover.
